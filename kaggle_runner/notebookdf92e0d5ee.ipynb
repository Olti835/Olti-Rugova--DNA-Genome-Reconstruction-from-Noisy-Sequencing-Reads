```python
import subprocess
import sys
from pathlib import Path

repo_url = "https://github.com/Olti835/Olti-Rugova--DNA-Genome-Reconstruction-from-Noisy-Sequencing-Reads"

project_root = Path("/kaggle/working/GenomeReconstructionProject_DAA")

if not project_root.exists():
    subprocess.run(
        ["git", "clone", repo_url, str(project_root)],
        check=True
    )
else:
    subprocess.run(
        ["git", "-C", str(project_root), "pull"],
        check=True
    )

sys.path.insert(0, str(project_root / "src"))

from genome_assembler import read_fastq

file_path = "/kaggle/input/datasets/oltirugova/dna-reads/SRR29154704.fastq"

reads = read_fastq(file_path)

print("Numri i reads:", len(reads))
print("Pesë reads e para:", reads[:5])
```
